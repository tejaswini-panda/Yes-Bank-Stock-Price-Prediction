#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value

from scipy.stats import pearsonr

correlation, p_value = pearsonr(df['Open'], df['Close'])

print("Pearson Correlation:", correlation)
print("P-Value:", p_value)
````

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

*   Alternative Hypothesis (H₁): There is a significant difference in the average Yes Bank closing price across different years.



#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value

from scipy.stats import f_oneway

yearly_groups = [
    group['Close'].values
    for _, group in df.groupby(df['Date'].dt.year)
]

f_statistic, p_value = f_oneway(*yearly_groups)

print("F-Statistic:", f_statistic)
print("P-Value:", p_value)
````

##### Which statistical test have you done to obtain P-Value?

Answer Here.

I performed a One-Way ANOVA test to obtain the p-value.

##### Why did you choose the specific statistical test?

Answer Here.

I chose the One-Way ANOVA test because the hypothesis compares the average closing prices across multiple years.
